### Ice Time Distribution & Minimum Threshold

Ice time is measured in seconds. Key percentiles:
- 25th percentile: ~242 minutes of 5on5 ice time
- Median: ~745 minutes
- Max: ~1,646 minutes

**Confidence threshold: 242 minutes (25th percentile)**

Players below this threshold will not be excluded but instead have their 
score pulled toward the league average using confidence weighting:

`final_score = (confidence × player_score) + ((1 - confidence) × league_average)`

Where confidence scales from 0 to 1 based on ice time relative to the threshold.
This means fringe players who barely played aren't penalised unfairly but also 
don't skew the rankings based on a tiny sample size.

### What Happens After a Giveaway?

We want to understand what typically follows a giveaway. We can do this by looking at the next event within the same game after each giveaway event.

```python
# Create an empty list to store what happens after each giveaway
next_event_types = []

# Loop through every giveaway
for idx, row in giveaways.iterrows():
    
    # Find all events in the same game that happened after this giveaway
    after = events[
        (events['game_id'] == row['game_id']) & 
        (events['sort_order'] > row['sort_order'])
    ]
    
    # Grab the very next event
    if len(after) > 0:
        next_event_types.append(after.iloc[0]['event_type'])
    else:
        next_event_types.append('none')
        
pd.Series(next_event_types).value_counts()
```

## Methodology & Metrics

### Overview
This project builds a composite puck management score for every NHL skater 
in the 2025-26 regular season. The goal is to move beyond simple counting 
statistics and build a context-adjusted, data-driven framework that fairly 
evaluates how well players manage the puck.

All analysis is restricted to 5on5 situations (situation code 1551) to ensure 
a level playing field since special teams situations distort puck management metrics 
since players are under fundamentally different levels of pressure.

---

### Data Sources

**MoneyPuck Season Level Skaters Data**
Season level aggregated statistics for all 939 NHL skaters in the 2025-26 
regular season. Filtered to 5on5 situations only. Provides clean, reliable 
counting statistics for giveaways, takeaways, zone starts and ice time.

**NHL API Play-by-Play Data**
Raw event level data pulled directly from the NHL's public API for all 1,312 
regular season games (413,859 total events). Each giveaway event includes exact 
x/y coordinates, zone, manpower situation, period, and timing. 

---

### Metrics

We settled on three distinct, non-redundant metrics that each measure a 
genuinely different dimension of puck management:

**Metric 1 — Giveaway Rate per 60**

How often a player turns the puck over per 60 minutes of 5on5 ice time. 
Lower is better. We initially considered a net turnover metric 
(takeaway rate - giveaway rate) but found that raw giveaway and takeaway 
counts correlate at r=0.83, meaning players involved in more puck battles 
tend to have both higher giveaways and takeaways. Including both separately 
would double count the same behaviour.

**Metric 2 — Giveaway Above Expected**

Our EDA showed that shots immediately follow 48% of all giveaways. However 
not all giveaways are equally dangerous — a turnover in the offensive zone 
corner is fundamentally different from one in front of your own net. 

For each giveaway we look at the next shot within 10 seconds and record its 
xGoal value (the probability it becomes a goal). We will then train a regression 
model using the giveaway's context — x/y coordinates, zone, period, time 
remaining, and score differential — to learn what the expected danger is for 
a giveaway in any given situation. 

A player's giveaway above expected score is the difference between the actual 
danger their giveaways caused and what was expected given the situations they 
were in. A negative score means their giveaways caused less danger than 
expected — good. A positive score means they turned it over in more dangerous 
spots than average — bad.

This directly addresses the key weakness of manual weighting systems where 
arbitrary values are assigned to different event types without data backing.

**Metric 3 — Defensive Zone Giveaway Rate (adjusted for zone starts)**

While giveaway above expected captures how dangerous turnovers are, it doesn't 
explicitly isolate how frequently a player turns it over in their own end. We 
calculate dzone giveaways per defensive zone shift start rather than per 60 
minutes. This adjusts for zone deployment — a player who starts 70% of shifts 
in the defensive zone will naturally face more opportunities to give it away 
there. Dividing by dzone shift starts makes this a fair comparison across all 
player types.

---

### Composite Score

Each metric is independently normalised using MAD (Median Absolute Deviation) 
z-scoring within position groups as forwards and defensemen are scored separately 
since they have fundamentally different roles and usage patterns. The MAD z-score forumla
is shown below: 

$$\text{MAD Z-Score} = 0.6745 \times \frac{x - \text{median}(x)}{\text{MAD}(x)}$$

A confidence weight is then applied based on each player's 5on5 ice time. 
Players below the 25th percentile threshold (242 minutes) have their score 
pulled toward the positional average rather than being excluded entirely:

The three normalised, confidence weighted metrics are then combined into a 
single composite puck management score. Equal weights are applied initially 
and adjusted based on correlation analysis and model performance.
